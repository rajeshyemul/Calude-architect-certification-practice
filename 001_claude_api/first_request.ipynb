# Making Your First Request to the Anthropic API

This notebook demonstrates how to configure your environment and make your first request to Claude using the Anthropic Python SDK.

### Step 2: Load Environment Variables & Initialize Client
Ensure you have a `.env` file containing:
```env
ANTHROPIC_API_KEY="your-api-key-here"
```

````python
import os
from dotenv import load_dotenv
from anthropic import Anthropic

# Load .env file
load_dotenv()

# Initialize Anthropic client (reads ANTHROPIC_BASE_URL if set in .env)
api_key = os.getenv("ANTHROPIC_API_KEY")
base_url = os.getenv("ANTHROPIC_BASE_URL")
client = Anthropic(api_key=api_key, base_url=base_url) if base_url else Anthropic(api_key=api_key)
model = os.getenv("ANTHROPIC_MODEL", "claude-3-5-sonnet-latest")
````

### Step 3: Making the Request (`client.messages.create`)
Key parameters:
- `model`: The model identifier
- `max_tokens`: Safety cap on generated output tokens
- `messages`: History of messages with roles (`user`, `assistant`)

````python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence."
        }
    ]
)
````

### Step 4: Extracting the Response
Access generated text via `message.content[0].text`.

````python
from anthropic.types import TextBlock

for block in message.content:
    if isinstance(block, TextBlock):
        print(block.text)
````
